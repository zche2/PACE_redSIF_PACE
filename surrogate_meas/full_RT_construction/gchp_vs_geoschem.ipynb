Compare GCHP and GEOS-Chem aerosol AOD variations

```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr

```

```python
geoschem_aod = xr.open_dataset('/home/zhe2/FraLab/PACE_redSIF_PACE/surrogate_meas/full_RT_construction/output_aerosol_profiles/geoschem_ocean_columns_n500.nc')
gchp_aod = xr.open_dataset('/home/zhe2/FraLab/PACE_redSIF_PACE/surrogate_meas/full_RT_construction/output_aerosol_profiles/gchp_ocean_columns_n500.nc')

geos_aod_tot_layer = geoschem_aod['AOD_total_layer'].values
gchp_aod_tot_layer = gchp_aod['AOD_total_layer'].values

geos_aod_tot = geos_aod_tot_layer.sum(axis=1)
gchp_aod_tot = gchp_aod_tot_layer.sum(axis=1)

geos_aod_min, geos_aod_max = np.min(geos_aod_tot), np.max(geos_aod_tot)
gchp_aod_min, gchp_aod_max = np.min(gchp_aod_tot), np.max(gchp_aod_tot)

print(f"GEOS-Chem AOD range: {geos_aod_min} to {geos_aod_max}")
print(f"GCHP AOD range: {gchp_aod_min} to {gchp_aod_max}")


```

```python
# sea salt aod
geos_ss_tot_layer = geoschem_aod['AOD_seasalt'].values
gchp_ss_tot_layer = gchp_aod['AOD_seasalt'].values

geos_ss_tot = geos_ss_tot_layer.sum(axis=1)
gchp_ss_tot = gchp_ss_tot_layer.sum(axis=1)

geos_ss_min, geos_ss_max = np.min(geos_ss_tot), np.max(geos_ss_tot)
gchp_ss_min, gchp_ss_max = np.min(gchp_ss_tot), np.max(gchp_ss_tot)

print(f"GEOS-Chem sea salt AOD range: {geos_ss_min} to {geos_ss_max}")
print(f"GCHP sea salt AOD range: {gchp_ss_min} to {gchp_ss_max}")
```
