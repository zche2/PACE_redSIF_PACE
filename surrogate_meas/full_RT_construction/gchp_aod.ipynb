### visualize GCHP AOD: change with wavelength

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
```

```python
file = "/home/zhe2/FraLab/PACE_redSIF_PACE/surrogate_meas/full_RT_construction/output_new_Lambertian_gchpio_finerwavnum0p05/rt_toa_ensemble_lambertian_gchpio.nc"
# read file
ds = xr.open_dataset(file)
# AOD
wavelength = ds.wavelength
aod = ds.aod

# plot
plt.figure(figsize=(10, 6))
plt.plot(wavelength, aod[::20, :].transpose())
plt.xlabel('Wavelength (nm)')
```
